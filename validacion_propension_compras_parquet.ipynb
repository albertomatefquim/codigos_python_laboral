## Revision de las key de ambas sabanas

```python
import numpy as np

keys_orig = set(df_prcom_orig['NUM_CLIENTE'])
keys_opt = set(df_prcom_opt['NUM_CLIENTE'])

keys_comunes = keys_orig.intersection(keys_opt)
solo_original = keys_orig - keys_opt
solo_optimizado = keys_opt - keys_orig

print(f"Clientes en ambos datasets: {len(keys_comunes):,}")
print(f"Clientes solo en 'original': {len(solo_original):,}")
print(f"Clientes solo en 'optimizado': {len(solo_optimizado):,}")
```

## Creamos codigo para detectar las diferencias

```python
import datacompy

compare = datacompy.PandasCompare(
    df1=df_prcom_orig[df_prcom_opt.columns],
    df2=df_prcom_opt,
    join_columns=['NUM_CLIENTE', 'ANTIGUEDAD_AHORRO','OFICINA_SERVICIO'],
    df1_name='original',
    df2_name='optimizado',
    abs_tol = 0.0001,
    rel_tol = 0
)

print(compare.report(sample_count = 100))
```

```python
df_diferencias = compare.all_mismatch(ignore_matching_cols=True)
```
