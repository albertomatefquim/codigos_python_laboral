```python
import pandas as pd
import numpy as np
```

```python
parquet_junio = 'C:/Users/E-LHPEREZ/Desktop/Backtest Yastas/Backtest/SabanaAnalitica_MalasPracticasYastas_24ago26_junio26.parquet'
df_junio = pd.read_parquet(parquet_junio)

parquet_julio = 'C:/Users/E-LHPEREZ/Desktop/Backtest Yastas/Backtest/SabanaAnalitica_MalasPracticasYastas_24ago26_julio26.parquet'
df_julio = pd.read_parquet(parquet_julio)
```

```python
def reporte_calidad_dataframe(df):
    """
    Analiza un DataFrame y devuelve un nuevo DataFrame con todas las métricas
    de calidad de datos agregadas por columna.
    """
    total_filas = df.shape[0]
    metricas = []

    for col in df.columns:
        # 1. Tipo de dato y conteos básicos
        tipo_dato = df[col].dtype
        nulos = df[col].isnull().sum()
        porc_nulos = round((nulos / total_filas) * 100, 2)
        unicos = df[col].nunique()
        porc_unicos = round((unicos / total_filas) * 100, 2)
        
        # Valores por defecto para variables no numéricas
        min_val = np.nan
        max_val = np.nan
        outliers_count = 0
        porc_outliers = 0.0

        # 2. Métricas y Outliers solo para columnas numéricas
        if pd.api.types.is_numeric_dtype(df[col]):
            min_val = df[col].min()
            max_val = df[col].max()
            
            # Cálculo IQR de Outliers
            col_sin_nulos = df[col].dropna()
            if len(col_sin_nulos) > 0:
                Q1 = col_sin_nulos.quantile(0.25)
                Q3 = col_sin_nulos.quantile(0.75)
                IQR = Q3 - Q1
                limite_inferior = Q1 - 1.5 * IQR
                limite_superior = Q3 + 1.5 * IQR
                
                outliers_count = ((df[col] < limite_inferior) | (df[col] > limite_superior)).sum()
                porc_outliers = round((outliers_count / total_filas) * 100, 2)

        # 3. Guardar registro por columna
        metricas.append({
            'columna': col,
            'tipo_dato': str(tipo_dato),
            'nulos_total': nulos,
            'nulos_porc': porc_nulos,
            'unicos_total': unicos,
            'unicos_porc': porc_unicos,
            'valor_min': min_val,
            'valor_max': max_val,
            'outliers_total': outliers_count,
            'outliers_porc': porc_outliers
        })

    # Convertimos a DataFrame
    df_reporte = pd.DataFrame(metricas)
    return df_reporte
```
